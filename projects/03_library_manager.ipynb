## Project brief

Create a small library that can add books, search titles, borrow available books, return them,
and produce a summary. `Book` owns information about one book; `Library` manages the collection.


## Step 1 · Model a book


```python
class Book:
    def __init__(self, title, author):
        if not title.strip() or not author.strip():
            raise ValueError("title and author are required")
        self.title = title.strip()
        self.author = author.strip()
        self.is_available = True

    def borrow(self):
        if not self.is_available:
            raise ValueError(f"{self.title} is already borrowed")
        self.is_available = False

    def return_book(self):
        self.is_available = True

    def to_dict(self):
        return {
            "title": self.title,
            "author": self.author,
            "is_available": self.is_available,
        }

    def __str__(self):
        status = "available" if self.is_available else "borrowed"
        return f"{self.title} by {self.author} ({status})"
```

## Step 2 · Manage a collection


```python
class Library:
    def __init__(self, name):
        self.name = name
        self.books = []

    def add_book(self, book):
        if not isinstance(book, Book):
            raise TypeError("book must be a Book instance")
        self.books.append(book)

    def find_by_title(self, search_text):
        search_text = search_text.casefold()
        return [book for book in self.books if search_text in book.title.casefold()]

    def available_books(self):
        return [book for book in self.books if book.is_available]

    def summary(self):
        available = len(self.available_books())
        return {"name": self.name, "books": len(self.books), "available": available}
```

## Step 3 · Use the library


```python
library = Library("Python Learning Library")
library.add_book(Book("Python Crash Course", "Eric Matthes"))
library.add_book(Book("Automate the Boring Stuff", "Al Sweigart"))
library.add_book(Book("Think Python", "Allen B. Downey"))

matches = library.find_by_title("python")
matches[0].borrow()

for book in library.books:
    print(book)
print(library.summary())
```

## Step 4 · Check normal and error paths


```python
assert library.summary() == {"name": "Python Learning Library", "books": 3, "available": 2}

try:
    matches[0].borrow()
except ValueError as error:
    print("Expected rule:", error)

matches[0].return_book()
assert len(library.available_books()) == 3
print("All checks passed.")
```
