## Project brief

Build a game that chooses a number from 1–100, allows a limited number of guesses, reports
“too high” or “too low,” and ends when the answer is correct or attempts run out.

In a notebook, keeping input separate from logic makes the game easy to test. We first build
`evaluate_guess`, then a game engine, and finally an optional interactive adapter.


## Step 1 · Evaluate one guess


```python
def evaluate_guess(guess, secret):
    if guess < secret:
        return "too low"
    if guess > secret:
        return "too high"
    return "correct"


assert evaluate_guess(25, 42) == "too low"
assert evaluate_guess(60, 42) == "too high"
assert evaluate_guess(42, 42) == "correct"
```

## Step 2 · Build the testable game engine


```python
def play_game(secret, guesses, max_attempts=7):
    if not 1 <= secret <= 100:
        raise ValueError("secret must be between 1 and 100")

    history = []
    for attempt, guess in enumerate(guesses[:max_attempts], start=1):
        if not 1 <= guess <= 100:
            history.append((attempt, guess, "outside range"))
            continue
        result = evaluate_guess(guess, secret)
        history.append((attempt, guess, result))
        if result == "correct":
            break
    return history
```

## Step 3 · Run a repeatable game


```python
secret = 42
sample_guesses = [50, 25, 40, 42]
history = play_game(secret, sample_guesses)

for attempt, guess, result in history:
    print(f"Attempt {attempt}: {guess} — {result}")

assert history[-1][2] == "correct"
```
