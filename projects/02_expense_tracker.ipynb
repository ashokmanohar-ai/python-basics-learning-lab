## Project brief

Build an expense tracker that validates entries, calculates totals by category, finds the
largest expense, and persists data as JSON. Currency is stored as a float for beginner
simplicity; production financial software should normally use `decimal.Decimal`.


## Step 1 · Create and validate records


```python
def create_expense(description, amount, category):
    description = description.strip()
    category = category.strip().title()
    amount = float(amount)
    if not description or not category:
        raise ValueError("description and category are required")
    if amount <= 0:
        raise ValueError("amount must be positive")
    return {"description": description, "amount": amount, "category": category}


expenses = [
    create_expense("Groceries", 42.50, "food"),
    create_expense("Bus pass", 30, "travel"),
    create_expense("Coffee", 3.25, "food"),
]
print(expenses)
```

## Step 2 · Calculate summaries


```python
def total_spent(expense_records):
    return sum(expense["amount"] for expense in expense_records)


def totals_by_category(expense_records):
    totals = {}
    for expense in expense_records:
        category = expense["category"]
        totals[category] = totals.get(category, 0) + expense["amount"]
    return totals


print(f"Total: £{total_spent(expenses):.2f}")
print(totals_by_category(expenses))
print("Largest:", max(expenses, key=lambda expense: expense["amount"]))
```

## Step 3 · Save and load JSON


```python
import json
from pathlib import Path
from tempfile import TemporaryDirectory


def save_expenses(expense_records, path):
    Path(path).write_text(json.dumps(expense_records, indent=2), encoding="utf-8")


def load_expenses(path):
    path = Path(path)
    if not path.exists():
        return []
    return json.loads(path.read_text(encoding="utf-8"))


temporary_workspace = TemporaryDirectory()
data_path = Path(temporary_workspace.name) / "expenses.json"
save_expenses(expenses, data_path)
restored = load_expenses(data_path)
assert restored == expenses
print(data_path.read_text(encoding="utf-8"))
```

## Step 4 · Produce a simple report


```python
def print_report(expense_records):
    print("EXPENSE REPORT")
    print("-" * 30)
    for category, amount in sorted(totals_by_category(expense_records).items()):
        print(f"{category:<15} £{amount:>8.2f}")
    print("-" * 30)
    print(f"{'Total':<15} £{total_spent(expense_records):>8.2f}")


print_report(restored)
temporary_workspace.cleanup()
```
