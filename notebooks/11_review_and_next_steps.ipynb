## Integrated example: analyse scores

This function combines validation, lists, dictionaries, loops, conditions, and exceptions.


```python
def analyse_scores(scores):
    if not scores:
        raise ValueError("scores cannot be empty")
    if any(not 0 <= score <= 100 for score in scores):
        raise ValueError("scores must be between 0 and 100")

    average = sum(scores) / len(scores)
    return {
        "count": len(scores),
        "lowest": min(scores),
        "highest": max(scores),
        "average": round(average, 2),
        "passed": sum(score >= 60 for score in scores),
    }


report = analyse_scores([72, 91, 58, 84, 66])
print(report)
```

## Assertions as self-checks

`assert` is useful while learning and in tests. It documents an expected result. Do not use
assertions to validate untrusted runtime input in production programs.


```python
assert analyse_scores([60])["passed"] == 1
assert analyse_scores([0, 100])["average"] == 50

try:
    analyse_scores([])
except ValueError as error:
    assert str(error) == "scores cannot be empty"

print("All self-checks passed.")
```
