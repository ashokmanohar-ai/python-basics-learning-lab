## Arithmetic operators

Python supports addition `+`, subtraction `-`, multiplication `*`, division `/`, floor
division `//`, remainder `%`, and exponentiation `**`. Parentheses make evaluation order clear.


```python
a, b = 17, 5
print("add:", a + b)
print("subtract:", a - b)
print("multiply:", a * b)
print("divide:", a / b)
print("floor divide:", a // b)
print("remainder:", a % b)
print("power:", a ** 2)
```

## Comparison and logical operators

Comparisons return `True` or `False`: `==`, `!=`, `<`, `<=`, `>`, `>=`. Combine them with
`and`, `or`, and `not`.


```python
age = 22
has_ticket = True
can_enter = age >= 18 and has_ticket

print("Can enter:", can_enter)
print("Needs help:", not can_enter)
print("Teen or senior:", age < 20 or age >= 65)
```

## Assignment operators

`+=`, `-=`, `*=`, and similar operators update a variable using its current value.


```python
score = 10
score += 5
score *= 2
score -= 4
print(score)
```

## Membership and identity

Use `in` to test membership. Use `is` mainly for identity checks such as `value is None`.
Do not use `is` as a replacement for `==`.


```python
skills = ["Python", "SQL", "Git"]
selected_skill = None

print("Python" in skills)
print("Java" not in skills)
print(selected_skill is None)
```

## Try it yourself: bill calculator

Calculate the total after a 10% discount, then check whether free delivery applies at £50.


```python
unit_price = 18.00
quantity = 3
discount_rate = 0.10

subtotal = unit_price * quantity
total = subtotal * (1 - discount_rate)
free_delivery = total >= 50
print(f"Total: £{total:.2f}; free delivery: {free_delivery}")
```
