## Exceptions are structured error signals

Common exceptions include `ValueError`, `TypeError`, `KeyError`, `IndexError`, and
`ZeroDivisionError`. Handle exceptions only when you can recover or add useful context.


```python
examples = [ValueError, TypeError, KeyError, IndexError, ZeroDivisionError]
print([exception.__name__ for exception in examples])
```

## `try` and specific `except` blocks

Put the risky operation in `try`. Catch specific exceptions rather than a broad `except`,
because unrelated programming errors should remain visible.


```python
def safe_divide(numerator_text, denominator_text):
    try:
        numerator = float(numerator_text)
        denominator = float(denominator_text)
        return numerator / denominator
    except ValueError:
        return "Please provide numbers."
    except ZeroDivisionError:
        return "The denominator cannot be zero."


for values in [("10", "2"), ("ten", "2"), ("10", "0")]:
    print(values, "->", safe_divide(*values))
```

## `else` and `finally`

`else` runs when no exception occurs. `finally` always runs, which is useful for cleanup.
Keep the `try` block narrow so it covers only operations expected to fail.


```python
try:
    result = int("42")
except ValueError:
    print("Conversion failed")
else:
    print("Converted:", result)
finally:
    print("Conversion attempt complete")
```

## Raising an exception

Use `raise` when a function receives invalid data and cannot produce a meaningful result.
A clear message makes the contract easy to understand.


```python
def percentage(part, whole):
    if whole <= 0:
        raise ValueError("whole must be greater than zero")
    return part / whole * 100


print(percentage(45, 60))
try:
    percentage(1, 0)
except ValueError as error:
    print("Expected validation:", error)
```

## Try it yourself

Write `parse_age(text)` that converts text to an integer, rejects negative ages, and returns
a helpful message for invalid input.


```python
def parse_age(text):
    try:
        age = int(text)
        if age < 0:
            raise ValueError("age cannot be negative")
        return age
    except ValueError as error:
        return f"Invalid age: {error}"


print([parse_age(value) for value in ["25", "-1", "unknown"]])
```
