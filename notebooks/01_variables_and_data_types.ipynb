## Variables: names that refer to values

Assignment uses `=`. Choose meaningful `snake_case` names. Python is case-sensitive, so
`age` and `Age` are different names.


```python
learner_name = "Maya"
lessons_completed = 2
completion_rate = 16.7
is_beginner = True

print(learner_name, lessons_completed, completion_rate, is_beginner)
```

## Core scalar data types

| Type | Example | Meaning |
|---|---|---|
| `str` | `"Python"` | Text |
| `int` | `42` | Whole number |
| `float` | `3.14` | Decimal number |
| `bool` | `True` | True/false value |
| `NoneType` | `None` | No value yet |

Use `type()` to inspect a value.


```python
examples = ["Python", 42, 3.14, True, None]
for value in examples:
    print(repr(value), "->", type(value).__name__)
```

## Strings and f-strings

Strings can use single or double quotes. An f-string places expressions inside `{}`.


```python
topic = "variables"
minutes = 25
progress = 3 / 12
summary = f"I studied {topic} for {minutes} minutes ({progress:.0%} complete)."
print(summary)
```

## Type conversion

User input often arrives as text. Convert it with `int()`, `float()`, or `str()` before
calculations. Conversion only works when the value has a valid format.


```python
age_text = "28"
age = int(age_text)
price = float("19.95")

print(age + 1)
print(price * 2)
print(str(age) + " years old")
```
