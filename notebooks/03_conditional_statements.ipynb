## Making decisions with `if`

An `if` block runs only when its condition is truthy. Indentation defines the block; four
spaces is the standard. `elif` checks another condition and `else` provides a fallback.


```python
temperature = 24

if temperature >= 30:
    advice = "Stay hydrated."
elif temperature >= 20:
    advice = "Pleasant weather."
else:
    advice = "Take a jacket."

print(advice)
```

## Ranges and condition order

Conditions are checked from top to bottom, and only the first matching branch runs. Put more
specific or higher-threshold conditions first.


```python
score = 84

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "Needs practice"

print(grade)
```

## Compound and nested conditions

Logical operators often keep conditions flat and readable. Nest a condition only when the
second decision truly depends on the first.


```python
username = "learner"
password_is_valid = True
account_is_active = True

if username and password_is_valid and account_is_active:
    print("Access granted")
else:
    print("Access denied")
```

## Conditional expressions

Use `value_if_true if condition else value_if_false` for a short, simple choice. Prefer a
normal `if` statement when the logic needs explanation.


```python
items = 4
label = "item" if items == 1 else "items"
print(f"{items} {label}")
```

## Try it yourself: ticket price

Set a ticket price: free under age 5, £8 from 5–17, £12 from 18–64, and £9 for 65+.
Test boundary ages such as 4, 5, 17, 18, 64, and 65.


```python
age = 17

if age < 5:
    price = 0
elif age < 18:
    price = 8
elif age < 65:
    price = 12
else:
    price = 9

print(f"Ticket price: £{price}")
```
