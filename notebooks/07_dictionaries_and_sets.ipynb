## Learning objectives

- Store key-value data in dictionaries
- Safely read, update, and iterate over mappings
- Use sets for uniqueness and set operations


## Dictionaries: values accessed by key

A dictionary maps unique, hashable keys to values. Use a clear key such as `"name"` instead
of remembering a numeric position.


```python
learner = {
    "name": "Maya",
    "level": "beginner",
    "completed": 6,
}
print(learner["name"])
print(learner.get("city", "Not provided"))
```

## Add, update, and remove entries

Assignment adds or updates a key. `update()` can merge several values. `pop()` removes a key
and returns its value. Membership with `in` checks keys.


```python
learner["city"] = "Bengaluru"
learner.update({"completed": 7, "active": True})
city = learner.pop("city")
print("Removed city:", city)
print("active" in learner, learner)
```

## Iterating over a dictionary

`.items()` provides key-value pairs, `.keys()` provides keys, and `.values()` provides values.


```python
for key, value in learner.items():
    print(f"{key}: {value}")
```

## Sets: unordered collections of unique values

Sets automatically remove duplicates. They support union `|`, intersection `&`, difference `-`,
and symmetric difference `^`. Create an empty set with `set()`, not `{}`.


```python
backend_skills = {"Python", "SQL", "APIs"}
qa_skills = {"Python", "Testing", "APIs"}

print("All:", backend_skills | qa_skills)
print("Shared:", backend_skills & qa_skills)
print("Backend only:", backend_skills - qa_skills)
```

## Try it yourself: word frequency

Count how often each word appears. Then use a set to print the unique words.


```python
words = "python makes practice practical python makes progress".split()
frequencies = {}

for word in words:
    frequencies[word] = frequencies.get(word, 0) + 1

print(frequencies)
print("Unique words:", sorted(set(words)))
```
