## `for` loops

A `for` loop visits each item in an iterable such as a string, list, or range. The loop
variable receives one value at a time.


```python
topics = ["variables", "conditions", "loops"]

for topic in topics:
    print("Practising:", topic)
```

## `range()` and `enumerate()`

`range(start, stop, step)` generates integers but excludes `stop`. `enumerate()` supplies
both an index and a value without manually managing a counter.


```python
print(list(range(2, 11, 2)))

for position, topic in enumerate(topics, start=1):
    print(f"{position}. {topic}")
```

## Accumulator pattern

Start with an empty or zero value, then update it during each iteration. This pattern can
calculate totals, build collections, and count matches.


```python
prices = [4.50, 8.00, 2.25]
total = 0

for price in prices:
    total += price

print(f"Total: £{total:.2f}")
```

## `while`, `break`, and `continue`

A `while` loop repeats while its condition stays true. Always update the condition to avoid
an infinite loop. `break` exits; `continue` skips to the next iteration.


```python
number = 0
collected = []

while number < 10:
    number += 1
    if number % 2 != 0:
        continue
    collected.append(number)
    if len(collected) == 3:
        break

print(collected)
```

## Try it yourself: multiplication table

Print the 7-times table from `7 × 1` to `7 × 10`. Then replace 7 with another number.


```python
table_for = 7
for multiplier in range(1, 11):
    print(f"{table_for} × {multiplier} = {table_for * multiplier}")
```
