## Why functions?

A function gives a name to reusable logic. Define it with `def`, place its body in an
indented block, and call it with parentheses. A small function should usually do one job.


```python
def greet(name):
    return f"Welcome, {name}!"


message = greet("Maya")
print(message)
```

## Parameters, arguments, and defaults

Parameters appear in the definition; arguments are values supplied at the call. A default
makes an argument optional. Keyword arguments make calls easier to read.


```python
def calculate_total(price, quantity=1, discount=0.0):
    subtotal = price * quantity
    return subtotal * (1 - discount)


print(calculate_total(10))
print(calculate_total(price=10, quantity=3, discount=0.10))
```

## Return values versus printing

`return` sends a result back to the caller, so it can be stored, tested, or combined. `print()`
only displays a value. A function without an explicit return produces `None`.


```python
def rectangle_area(width, height):
    return width * height


area = rectangle_area(4, 6)
print("Area:", area)
assert area == 24
```

## Docstrings, type hints, and scope

A docstring explains purpose and behaviour. Type hints communicate expected types but do not
automatically enforce them. Variables created inside a function are local to that function.


```python
def celsius_to_fahrenheit(celsius: float) -> float:
    '''Convert a Celsius temperature to Fahrenheit.'''
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit


print(celsius_to_fahrenheit(25))
print(celsius_to_fahrenheit.__doc__)
```
