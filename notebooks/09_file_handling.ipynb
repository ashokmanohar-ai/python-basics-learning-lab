## Paths and context managers

`pathlib.Path` represents file-system paths. `with` automatically closes a file even if an
error occurs. This notebook uses a temporary directory, so its examples do not leave files behind.


```python
from pathlib import Path
from tempfile import TemporaryDirectory

workspace = TemporaryDirectory()
workdir = Path(workspace.name)
print("Temporary workspace:", workdir)
```

## Write and read text

`write_text()` and `read_text()` are convenient for small files. Specify UTF-8 explicitly.
Opening with mode `"a"` appends instead of replacing content.


```python
notes_path = workdir / "notes.txt"
notes_path.write_text("Variables\nLoops\nFunctions\n", encoding="utf-8")

with notes_path.open("a", encoding="utf-8") as file:
    file.write("Files\n")

print(notes_path.read_text(encoding="utf-8"))
```

## CSV data

CSV is a common tabular format. `csv.DictWriter` and `csv.DictReader` let us work with column
names instead of positions.


```python
import csv

csv_path = workdir / "scores.csv"
rows = [{"name": "Maya", "score": 88}, {"name": "Leo", "score": 94}]

with csv_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

with csv_path.open(encoding="utf-8") as file:
    loaded_rows = list(csv.DictReader(file))

print(loaded_rows)
```

## JSON data

JSON stores dictionaries, lists, strings, numbers, booleans, and null values. Use `dump()`
to write and `load()` to read. JSON object keys must be strings.


```python
import json

profile_path = workdir / "profile.json"
profile = {"name": "Maya", "skills": ["Python", "Git"], "active": True}

with profile_path.open("w", encoding="utf-8") as file:
    json.dump(profile, file, indent=2)

with profile_path.open(encoding="utf-8") as file:
    loaded_profile = json.load(file)

print(loaded_profile)
```

## Handling missing files

Check `.exists()` when absence is expected, or catch `FileNotFoundError` when the read itself
is the natural check.


```python
missing_path = workdir / "missing.txt"
try:
    missing_path.read_text(encoding="utf-8")
except FileNotFoundError:
    print(f"No file found at {missing_path.name}; using an empty value.")
```

## Try it yourself

Create a text file containing three goals, read it, and print each goal with a number. Then
try the same task with JSON.


```python
goals_path = workdir / "goals.txt"
goals = ["Finish the course", "Build a project", "Share it on GitHub"]
goals_path.write_text("\n".join(goals), encoding="utf-8")

for index, goal in enumerate(goals_path.read_text(encoding="utf-8").splitlines(), start=1):
    print(f"{index}. {goal}")

workspace.cleanup()
```
