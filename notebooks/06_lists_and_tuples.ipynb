## Lists: ordered, mutable collections

List indexes start at 0; negative indexes count from the end. Slices use `[start:stop:step]`
and exclude the stop index.


```python
languages = ["Python", "JavaScript", "SQL", "Java"]
print(languages[0])
print(languages[-1])
print(languages[1:3])
```

## Updating and copying lists

Lists can change after creation. `append()` adds one item, `extend()` adds many, `insert()`
adds at a position, and `remove()` deletes by value. Use `.copy()` when an independent
shallow copy is needed.


```python
skills = ["Python", "Git"]
skills.append("SQL")
skills.extend(["Testing", "APIs"])
skills.remove("Git")

backup = skills.copy()
skills[0] = "Advanced Python"
print("Current:", skills)
print("Backup:", backup)
```

## Sorting and useful built-ins

`sorted()` returns a new list; `.sort()` changes the original. `len()`, `min()`, `max()`, and
`sum()` provide common summaries.


```python
scores = [72, 91, 65, 88]
print("Sorted:", sorted(scores))
print("Original:", scores)
print("Average:", sum(scores) / len(scores))
```

## List comprehensions

A comprehension builds a list from an iterable: `[expression for item in iterable if condition]`.
Use a normal loop if the transformation becomes difficult to read.


```python
squares = [number ** 2 for number in range(1, 6)]
even_squares = [value for value in squares if value % 2 == 0]
print(squares)
print(even_squares)
```

## Tuples: ordered, immutable collections

Tuples cannot be changed after creation, which suits fixed records and multiple return values.
Tuple unpacking assigns each item to a name.


```python
point = (12, 8)
x, y = point
print(f"x={x}, y={y}")

single_item_tuple = ("Python",)
print(type(single_item_tuple).__name__)
```

## Try it yourself

Start with five temperatures. Print the highest, lowest, and average, then create a new list
containing only values above the average.


```python
temperatures = [18, 22, 25, 19, 27]
average = sum(temperatures) / len(temperatures)
above_average = [value for value in temperatures if value > average]
print(min(temperatures), max(temperatures), average, above_average)
```
