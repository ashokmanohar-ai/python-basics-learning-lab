## Classes and objects

A class is a blueprint; an object is one instance. `__init__` initializes a new object, and
`self` refers to that object inside its methods.


```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def describe(self):
        return f"{self.name} is {self.age} years old."


pet = Dog("Pixel", 3)
print(pet.describe())
```

## State and behaviour

Attributes hold state; methods define behaviour. The object can update its own state while
enforcing simple rules.


```python
class BankAccount:
    def __init__(self, owner, opening_balance=0.0):
        self.owner = owner
        self.balance = float(opening_balance)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("insufficient funds")
        self.balance -= amount
        return self.balance


account = BankAccount("Maya", 100)
account.deposit(25)
account.withdraw(40)
print(account.balance)
```

## Helpful string representation

`__str__` controls the friendly text produced by `str(object)` and `print(object)`.


```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"{self.title} by {self.author}"


book = Book("Automate the Boring Stuff", "Al Sweigart")
print(book)
```

## Inheritance and composition

Inheritance models an “is-a” relationship; composition models a “has-a” relationship. Prefer
the simplest design. Beginners should first master small classes before building hierarchies.


```python
class Employee:
    def __init__(self, name):
        self.name = name

    def role(self):
        return "Employee"


class Tester(Employee):
    def role(self):
        return "Quality Engineer"


person = Tester("Maya")
print(person.name, "-", person.role())
```

## Try it yourself

Create a `Rectangle` class with width and height attributes plus `area()` and `perimeter()`
methods. Add validation so both dimensions must be positive.


```python
class Rectangle:
    def __init__(self, width, height):
        if width <= 0 or height <= 0:
            raise ValueError("dimensions must be positive")
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


rectangle = Rectangle(4, 6)
print(rectangle.area(), rectangle.perimeter())
```
