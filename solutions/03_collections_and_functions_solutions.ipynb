## 1. Clean names


```python
def clean_names(names):
    return [name.strip().title() for name in names if name.strip()]

assert clean_names(["  maya", "LEO ", ""]) == ["Maya", "Leo"]
```

## 2. Word counts


```python
def word_counts(text):
    counts = {}
    for word in text.lower().split():
        counts[word] = counts.get(word, 0) + 1
    return counts

print(word_counts("Python practice makes Python progress"))
```

## 3. Unique shared skills


```python
def shared_skills(first, second):
    return sorted(set(first) & set(second))

assert shared_skills(["Python", "SQL"], ["Git", "Python"]) == ["Python"]
```

## 4. Student summary


```python
def summarise_students(records, pass_mark=60):
    if not records:
        raise ValueError("records cannot be empty")
    top_name, top_score = max(records, key=lambda record: record[1])
    return {
        "average": sum(score for _, score in records) / len(records),
        "top_student": top_name,
        "top_score": top_score,
        "pass_count": sum(score >= pass_mark for _, score in records),
    }

print(summarise_students([("Maya", 88), ("Leo", 94), ("Sam", 57)]))
```

## 5. Group by first letter


```python
def group_names(names):
    grouped = {}
    for name in names:
        grouped.setdefault(name[0].upper(), []).append(name)
    return {letter: sorted(values) for letter, values in sorted(grouped.items())}

print(group_names(["Maya", "Max", "Leo", "Lina", "Sam"]))
```
