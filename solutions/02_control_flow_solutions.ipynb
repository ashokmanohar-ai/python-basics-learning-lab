## 1. Number classifier


```python
number = -4
sign = "positive" if number > 0 else "negative" if number < 0 else "zero"
parity = "even" if number % 2 == 0 else "odd"
print(sign, parity)
```

## 2. Leap year


```python
year = 2028
is_leap_year = year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
assert is_leap_year
print(is_leap_year)
```

## 3. FizzBuzz


```python
for number in range(1, 31):
    if number % 15 == 0:
        label = "FizzBuzz"
    elif number % 3 == 0:
        label = "Fizz"
    elif number % 5 == 0:
        label = "Buzz"
    else:
        label = str(number)
    print(label)
```

## 4. Sum selected values


```python
total = sum(number for number in range(1, 101) if number % 3 == 0 or number % 5 == 0)
assert total == 2418
print(total)
```

## 5. Password attempts


```python
expected = "python123"
attempts = ["password", "python", "python123", "ignored"]
for attempt_number, attempt in enumerate(attempts[:3], start=1):
    if attempt == expected:
        print(f"Access granted on attempt {attempt_number}")
        break
else:
    print("Access denied")
```
