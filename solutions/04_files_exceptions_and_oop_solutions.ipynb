```python
from pathlib import Path
from tempfile import TemporaryDirectory
import json

temporary_workspace = TemporaryDirectory()
workdir = Path(temporary_workspace.name)
```

## 1. Safe integer


```python
def safe_int(text, default=0):
    try:
        return int(text)
    except (TypeError, ValueError):
        return default

assert safe_int("42") == 42
assert safe_int("unknown", -1) == -1
```

## 2. Save and load tasks


```python
tasks = [{"title": "Study Python", "done": False}]
tasks_path = workdir / "tasks.json"
tasks_path.write_text(json.dumps(tasks, indent=2), encoding="utf-8")
loaded_tasks = json.loads(tasks_path.read_text(encoding="utf-8"))
assert loaded_tasks == tasks
```

## 3–4. Product and inventory report


```python
class Product:
    def __init__(self, name, price, stock):
        if price < 0 or stock < 0:
            raise ValueError("price and stock cannot be negative")
        self.name, self.price, self.stock = name, float(price), int(stock)

    def sell(self, quantity):
        if quantity <= 0:
            raise ValueError("quantity must be positive")
        if quantity > self.stock:
            raise ValueError("insufficient stock")
        self.stock -= quantity

    def to_dict(self):
        return {"name": self.name, "price": self.price, "stock": self.stock}

    def __str__(self):
        return f"{self.name}: £{self.price:.2f} × {self.stock}"


products = [Product("Notebook", 2.5, 10), Product("Pen", 1.2, 20)]
products[0].sell(2)
inventory_value = sum(product.price * product.stock for product in products)
inventory_path = workdir / "inventory.json"
inventory_path.write_text(
    json.dumps([product.to_dict() for product in products], indent=2), encoding="utf-8"
)
print(f"Inventory value: £{inventory_value:.2f}")
print(inventory_path.read_text(encoding="utf-8"))
temporary_workspace.cleanup()
```
