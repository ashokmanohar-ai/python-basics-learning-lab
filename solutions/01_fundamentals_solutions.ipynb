## 1. Personal introduction


```python
name, city, learning_goal = "Maya", "Bengaluru", "build useful programs"
introduction = f"I am {name} from {city}, and I want to {learning_goal}."
print(introduction)
```

## 2. Rectangle calculator


```python
width, height = 8, 5
area = width * height
perimeter = 2 * (width + height)
assert (area, perimeter) == (40, 26)
print(area, perimeter)
```

## 3. Temperature conversion


```python
celsius = 30
fahrenheit = (celsius * 9 / 5) + 32
assert fahrenheit == 86
print(fahrenheit)
```

## 4. Time conversion


```python
total_seconds = 7385
hours = total_seconds // 3600
remainder = total_seconds % 3600
minutes = remainder // 60
seconds = remainder % 60
assert (hours, minutes, seconds) == (2, 3, 5)
print(hours, minutes, seconds)
```

## 5. Shopping bill


```python
price, quantity, discount_rate = 24.99, 3, 0.15
subtotal = price * quantity
discount = subtotal * discount_rate
final_total = subtotal - discount
print(f"Subtotal £{subtotal:.2f}; discount £{discount:.2f}; total £{final_total:.2f}")
```
